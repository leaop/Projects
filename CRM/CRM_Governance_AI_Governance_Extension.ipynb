# CRM Access Governance & Customer Data Protection

11. Create AI risk scoring and classification.
12. Define monitoring KPIs.
13. Map the project to NIST AI RMF concepts.
14. Prepare the project for final GitHub documentation and portfolio packaging.

> **Important:** this is a simulated governance exercise. The controls are project proposals and should not be treated as universal legal, regulatory, or security requirements.

# 8. AI Risk Assessment Model

For the portfolio prototype, each risk is assessed using:

- Likelihood: 1–5
- Impact: 1–5
- Risk Score = Likelihood × Impact

Proposed interpretation:

- 1–4 → LOW
- 5–9 → MEDIUM
- 10–15 → HIGH
- 16–25 → CRITICAL


````python
ai_risk_register = pd.DataFrame([
    ["AI-RISK-01", 3, 5],
    ["AI-RISK-02", 3, 5],
    ["AI-RISK-03", 4, 4],
    ["AI-RISK-04", 3, 4],
    ["AI-RISK-05", 3, 5],
    ["AI-RISK-06", 2, 5],
    ["AI-RISK-07", 3, 4],
    ["AI-RISK-08", 3, 4],
    ["AI-RISK-09", 2, 4],
    ["AI-RISK-10", 4, 3],
    ["AI-RISK-11", 2, 4],
    ["AI-RISK-12", 2, 5]
], columns=["Risk_ID", "Likelihood", "Impact"])

ai_risk_register["Risk_Score"] = (
    ai_risk_register["Likelihood"] *
    ai_risk_register["Impact"]
)

def classify_ai_risk(score):
    if score <= 4:
        return "LOW"
    elif score <= 9:
        return "MEDIUM"
    elif score <= 15:
        return "HIGH"
    return "CRITICAL"

ai_risk_register["Risk_Level"] = (
    ai_risk_register["Risk_Score"]
    .apply(classify_ai_risk)
)

ai_risk_register = ai_risk_register.merge(
    ai_risk_taxonomy,
    on="Risk_ID",
    how="left"
)

display(
    ai_risk_register[
        [
            "Risk_ID",
            "Risk_Category",
            "Likelihood",
            "Impact",
            "Risk_Score",
            "Risk_Level"
        ]
    ].sort_values("Risk_Score", ascending=False)
)
````

# 19. Risk-to-Control Coverage

This check verifies whether identified AI risks have at least one mapped control.


````python
risk_ids = set(ai_risk_taxonomy["Risk_ID"])

mapped_risks = set()

for value in ai_control_catalog["Related_Risk_ID"]:
    for risk_id in risk_ids:
        if risk_id in value:
            mapped_risks.add(risk_id)

coverage = (
    len(mapped_risks) /
    len(risk_ids) *
    100
)

risk_control_coverage = pd.DataFrame({
    "Metric": [
        "Identified AI Risks",
        "Risks with Mapped Controls",
        "Risk-to-Control Coverage %"
    ],
    "Value": [
        len(risk_ids),
        len(mapped_risks),
        coverage
    ]
})

display(risk_control_coverage.round(2))
````

# 20. NIST AI RMF Mapping

This project uses the NIST AI Risk Management Framework as a **conceptual reference**, not as a formal certification claim.

The AI RMF Core is organized around four functions:

### GOVERN
Establish governance structure, responsibilities, policies, and accountability.

Project examples:
- AI system inventory;
- AI RACI;
- governance control catalog;
- review cadence;
- ownership.

### MAP
Understand the AI context, use case, affected data, users, and potential impacts.

Project examples:
- CRM AI use-case definition;
- data-access architecture;
- AI risk taxonomy;
- actor model;
- action classification.

### MEASURE
Assess and monitor AI risk.

Project examples:
- likelihood × impact scoring;
- risk levels;
- control coverage;
- AI governance KPIs;
- incident categories.

### MANAGE
Prioritize and respond to AI risks.

- human oversight;

- action restrictions;
