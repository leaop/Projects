# 2. Governance Operating Model Principles

A governance operating model should answer four questions:

1. **Who owns the decision?**
2. **Who maintains the control?**
3. **Who executes the process?**
4. **Who must be informed or consulted?**

This project uses four structural layers:

- **Accountability**
- **Stewardship**
- **Execution**
- **Oversight**

# 4. Data Owner vs. Data Steward

These two roles are frequently confused.

### Data Owner
- accountable for the data domain;
- approves business definitions;
- approves criticality and major policy decisions;
- accepts or rejects risk;
- sponsors remediation.

### Data Steward
- maintains metadata;
- monitors data quality;
- coordinates remediation;
- documents definitions;
- tracks issues;
- escalates unresolved problems.

The Owner is primarily **accountable**.  
The Steward is primarily **operationally responsible** for governance maintenance.

```text

      v

      |
```
